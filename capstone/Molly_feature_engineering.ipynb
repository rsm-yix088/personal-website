## 3. User Behavior Features

```python
display(user_df.head())
```

```python
user_df["fav_platform"].value_counts()
```

```python
user_df["n_platforms_used"].value_counts()
```

```python
# mobile_order_ratio
mobile_orders_df = (
    df_order_level[df_order_level["Platform"].isin(["ios", "android"])]
    .groupby("UserId")["OrderId"]
    .nunique()
    .reset_index(name="mobile_orders")
)
mobile_orders_df
# web_order_ratio
web_orders_df = (
    df_order_level[df_order_level["Platform"] == "web"]
    .groupby("UserId")["OrderId"]
    .nunique()
    .reset_index(name="web_orders")
)
# admin
admin_orders_df = (
    df_order_level[df_order_level["Platform"] == "admin"]
    .groupby("UserId")["OrderId"]
    .nunique()
    .reset_index(name="admin_orders")
)
# total
total_orders_df = (
    df_order_level.groupby("UserId")["OrderId"]
    .nunique()
    .reset_index(name="total_orders")
)
platform_df = (
    total_orders_df.merge(mobile_orders_df, on="UserId", how="left")
    .merge(web_orders_df, on="UserId", how="left")
    .merge(admin_orders_df, on="UserId", how="left")
)
# replace NAN as 0
platform_df[["mobile_orders", "web_orders", "admin_orders"]] = platform_df[
    ["mobile_orders", "web_orders", "admin_orders"]
].fillna(0)
display(platform_df)
```

```python
# platform_df["web_orders"] = platform_df["web_orders"] + platform_df["admin_orders"]
```

```python
# calculate ratio for different category(web, mobile, admin)
# mobile ratio(ios, android)
platform_df["mobile_order_ratio"] = (
    platform_df["mobile_orders"] / platform_df["total_orders"]
)
# web ratio
platform_df["web_order_ratio"] = platform_df["web_orders"] / platform_df["total_orders"]
# admin ratio
platform_df["admin_order_ratio"] = (
    platform_df["admin_orders"] / platform_df["total_orders"]
)
display(platform_df)
```

```python
# merge platform_df back to user_df
user_df = user_df.merge(
    platform_df[
        ["UserId", "mobile_order_ratio", "web_order_ratio", "admin_order_ratio"]
    ],
    on="UserId",
    how="left",
)
# calculate is_mobile_only
user_df["is_mobile_only"] = (
    (user_df["fav_platform"].isin(["ios", "android"]))
    & (user_df["n_platforms_used"] == 1)
).astype(int)
display(user_df)
```

## 4. Cuisine Affinity Features

Cuisine affinity features capture a user’s preference distribution across different cuisine types (e.g., Italian, Chinese, Fast Food). Instead of treating all orders equally, these features quantify what kinds of food a user tends to order, typically as proportions or counts.

The main purpose is to better represent user taste and behavioral patterns, which are not captured by standard metrics like frequency or monetary value. For example, two users may have the same number of orders, but one consistently orders fast food while another prefers premium or niche cuisines—this reflects very different customer profiles.

From a modeling and segmentation perspective, cuisine affinity helps:

- Improve customer segmentation by grouping users with similar preferences
- Enhance recommendation systems by identifying likely cuisine interests
- Support targeted marketing, such as cuisine-specific promotions or partnerships
- Reveal behavioral insights, such as whether certain cuisines are associated with higher spending or retention
